```python
import pandas as pd
df = pd.read_pickle("databases\data.pkl")
```

```python
from image_features.duplicate_detection import cluster_duplicate_images
dp = cluster_duplicate_images(df)
```

```python
def create_duplicate_backup(row, destination_folder):
    
    all_files = [x['filepath'] for x in row[0]]
    keep_file = [x['filepath'] for x in row[0] if x["bluriness"]==max([x["bluriness"] for x in row[0]])][0]
    all_files.remove(keep_file)
    import os
    import shutil
    os.makedirs(rf"{destination_folder}\{row["duplicated"]}", exist_ok=True)

    for file_path in all_files:
        destination_file = rf"{destination_folder}\{row["duplicated"]}\{file_path.split("\\")[-1]}"
        if os.path.exists(file_path):
            shutil.copy(file_path, destination_file)
            print(f"Moved: {file_path}")
        else:
            print(f"Not found: {file_path}")

    # Copy the extra file
    if os.path.exists(keep_file):
        destination_file = rf"{destination_folder}\{row["duplicated"]}\k_{keep_file.split("\\")[-1]}"
        shutil.copy(keep_file, destination_file)
        print(f"Copied: {keep_file}")
    else:
        print(f"Not found: {keep_file}")
```

```python
destination_folder = r"C:\Users\prabh\MyData\samarth_pics\ProcessedImages"
for index, row in dp.iterrows():
    create_duplicate_backup(row, destination_folder)
```
